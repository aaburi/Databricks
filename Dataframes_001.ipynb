```python
#%%capture
# file path /Volumes/demo/sample_data/samples/customer.csv

df = spark.read.csv("/Volumes/demo/sample_data/samples/customer.csv", header=True, inferSchema=True)
#display(df) for a better view 
#df.show()
df.write.mode("overwrite").saveAsTable("demo.sample_data.customer")
df.describe("customer_id","email","registration_date","gender").show()
df.printSchema()
```

```python
%skip
```

```python
from pyspark.sql.functions import min, max

df.select(
    min("registration_date").alias("min_registration_date"),
    max("registration_date").alias("max_registration_date")
).show()
```

```python
df.withColumn(
    "registration_date_str",
    df.registration_date.cast("string")
).describe(
    "customer_id", "email", "registration_date_str", "gender"
).show()
```
